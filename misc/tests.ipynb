```python
import jax.numpy as jnp
```

```python
def get_min_max(center: list, lengths: list):
    cen = jnp.array(center)
    lengths = jnp.array(lengths)
    dl = lengths / 2.0
    min_p = (cen - dl).tolist()
    max_p = (cen + dl).tolist()
    return [round(v, 3) for v in min_p], [round(v, 3) for v in max_p]

def get_center_lengths(min_p: list, max_p: list):
    min_arr = jnp.array(min_p)
    max_arr = jnp.array(max_p)
    
    # Center is the midpoint between min and max
    center = (max_arr + min_arr) / 2.0
    
    # Lengths are the difference between max and min
    lengths = max_arr - min_arr
    
    return [round(v, 3) for v in center.tolist()], [round(v, 3) for v in lengths.tolist()]
```

```python
wb_pos_min = jnp.array([-0.5, -0.5, 0.0])
wb_pos_max = jnp.array([0.75, 0.5, 1.0])

get_center_lengths(wb_pos_min,wb_pos_max)

# get_center_lengths([0.255, 0.0, 0.220],[0.55, 0.06, 0.479])
```
