```python
def evaluate_test_case(test_id, actual_model):
    """
    Compares the LLM output against the true answer and prints the results.
    """
    prompt_text = prompts[test_id]
    expected_dict = expected_answers[test_id]
    
    # Convert the Pydantic model to a standard Python dictionary for comparison
    actual_dict = actual_model.model_dump()
    
    # Check if the generated output matches the expected output exactly
    is_correct = (actual_dict == expected_dict)
    
    print(f"\n================ TEST CASE {test_id} ================")
    print(f"Question:    {prompt_text}")
    print("-" * 50)
    print(f"Ollama Ans:  {actual_dict}")
    print(f"True Answer: {expected_dict}")
    print("-" * 50)
    
    if is_correct:
        print("Result:      ✅ CORRECT (Outputs match exactly)")
    else:
        print("Result:      ❌ INCORRECT (Outputs do not match)")
        
    return is_correct

def evaluate_dumped_case(test_id, actual_dict):
    """
    Compares the LLM output against the true answer and prints the results.
    """
    prompt_text = prompts[test_id]
    expected_dict = expected_answers[test_id]
    
    # Convert the Pydantic model to a standard Python dictionary for comparison
    # actual_dict = actual_model.model_dump()
    
    # Check if the generated output matches the expected output exactly
    is_correct = (actual_dict == expected_dict)
    
    print(f"\n================ TEST CASE {test_id} ================")
    print(f"Question:    {prompt_text}")
    print("-" * 50)
    print(f"Ollama Ans:  {actual_dict}")
    print(f"True Answer: {expected_dict}")
    print("-" * 50)
    
    if is_correct:
        print("Result:      ✅ CORRECT (Outputs match exactly)")
    else:
        print("Result:      ❌ INCORRECT (Outputs do not match)")
        
    return is_correct

# evaluate_test_case(30,barrier)
```
