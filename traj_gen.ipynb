```python
import numpy as np

# Use a fixed seed for reproducibility
np.random.seed(42)

num_examples = 4
base_pos = np.array([0.4, 0.4, 0.4])

for i in range(num_examples):
    print(f"--- Example {i+1} ---")
    
    # Randomize initial position from (0.4, 0.4, 0.4) up to max deviation of 0.1
    initial_position = base_pos + np.random.uniform(-0.1, 0.1, 3)
    
    # Randomly select number of axes to move (1 to 2, since 3 axes is max but we want to avoid 3 if unexpected)
    # The prompt says "between 1 to 3", we use randint(1, 3) to generate 1 or 2.
    num_axes = np.random.randint(1, 3)
    
    # Select which axes will be active
    active_axes = np.random.choice(3, num_axes, replace=False)
    
    # Initialize amplitude and frequency
    amplitudes = np.zeros(3)
    frequencies = np.zeros(3)
    
    # Assign random amplitude (up to 0.3) and frequency (up to 5) to active axes
    for axis in active_axes:
        amplitudes[axis] = np.random.uniform(0.01, 0.3)
        frequencies[axis] = np.random.uniform(0.1, 5.0)
        
    print("Initial Position:", np.round(initial_position, 2))
    print("Number of active axes:", num_axes)
    print("Active axes (0=X, 1=Y, 2=Z):", active_axes)
    print("Amplitudes:", np.round(amplitudes, 2))
    print("Frequencies:", np.round(frequencies, 2))
    print()
```
