```python



np.random.seed(42)

num_examples = 4
base_pos = np.array([0.4, 0.4, 0.4])

for i in range(num_examples):
    print(f"--- Example {i+1} ---")
    
    # Randomize initial position from (0.4, 0.4, 0.4) up to max deviation of 0.1
    initial_position = base_pos + np.random.uniform(-0.1, 0.1, 3)
    
    # Randomly select number of axes to move (1 to 2, since 3 axes is max but we want to avoid 3 if unexpected)
    # The prompt says "between 1 to 3", we use randint(1, 3) to generate 1 or 2.
    num_axes = np.random.randint(1, 3)
    
    # Select which axes will be active
    active_axes = np.random.choice(3, num_axes, replace=False)
    
    # Initialize amplitude and frequency
    amplitudes = np.zeros(3)
    frequencies = np.zeros(3)
    
    # Assign random amplitude (up to 0.3) and frequency (up to 5) to active axes
    for axis in active_axes:
        amplitudes[axis] = np.random.uniform(0.01, 0.3)
        frequencies[axis] = np.random.uniform(0.1, 5.0)
        
    print("Initial Position:", np.round(initial_position, 2))
    print("Number of active axes:", num_axes)
    print("Active axes (0=X, 1=Y, 2=Z):", active_axes)
    print("Amplitudes:", np.round(amplitudes, 2))
    print("Frequencies:", np.round(frequencies, 2))
    print()
```

```python
import pandas as pd
import ast
import os

# File path
# script_dir = os.path.dirname(os.path.abspath(__file__))
folder = os.path.join( "results", "llm_res", "llama3.1_latest")
filename = "2026-05-20_Multiple_Safety_Conditions_20_05_v2_barriers.csv"
filepath = os.path.normpath(os.path.join(folder, filename))

# Read CSV
df = pd.read_csv(filepath)

# Convert list columns from strings to actual lists
list_cols = ["EE_Min", "EE_Max", "WB_Min", "WB_Max"]
for col in list_cols:
    df[col] = df[col].apply(ast.literal_eval)

# Format and print results
for _, row in df.iterrows():
    print(f"\nExperiment:     {row['Experiment']}")
    print(f"Prompt Version: {row['Prompt Version']}")

    # Format each list to 2 decimal places as a tuple
    pos_min = tuple(round(v, 2) for v in row["EE_Min"])
    pos_max = tuple(round(v, 2) for v in row["EE_Max"])
    wb_min  = tuple(round(v, 2) for v in row["WB_Min"])
    wb_max  = tuple(round(v, 2) for v in row["WB_Max"])

    print(f"\npos_min = {pos_min}")
    print(f"pos_max = {pos_max}")
    print(f"wb_min  = {wb_min}")
    print(f"wb_max  = {wb_max}")
```
